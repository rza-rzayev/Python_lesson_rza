```python
n = input("Enter a number: ")
n = int(n)

sum = 0
div_by_3 = 0

for i in range(1, n+1):
    if i%2 == 0:
        sum = sum + i

    if i%3 == 0 and i%5 == 0:
        div_by_3 += 1
        print(" 3 5")

    elif i%3 == 0:
        div_by_3 += 1
        print("3")

    elif i%5 == 0:
        print("5")

    else:
        print(i)

print("Sum of even numbers:", sum)
print("The number of nunber divided by 3:", div_by_3)
    
```
